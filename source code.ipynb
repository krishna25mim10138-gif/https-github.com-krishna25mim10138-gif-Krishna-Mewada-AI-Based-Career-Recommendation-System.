```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import tkinter as tk
from tkinter import messagebox

# -----------------------------
# Sample dataset (replace with CSV if needed)
# -----------------------------
data = pd.DataFrame({
    "math": [90, 70, 60, 85],
    "coding": [95, 80, 65, 70],
    "communication": [60, 85, 90, 75],
    "career": ["Software Engineer", "Data Scientist", "HR Manager", "UI/UX Designer"]
})

# -----------------------------
# ML Model Class
# -----------------------------
class CareerModel:
    def __init__(self):
        self.clf = DecisionTreeClassifier()

    def train(self, data):
        X = data[["math", "coding", "communication"]]
        y = data["career"]
        self.clf.fit(X, y)

    def recommend(self, student):
        df = pd.DataFrame([student])
        return self.clf.predict(df)[0]

# -----------------------------
# Command Line Execution
# -----------------------------
def run_cli():
    model = CareerModel()
    model.train(data)

    print("=== Career Recommendation System (CLI Mode) ===")
    math_score = int(input("Enter Math Score: "))
    coding_score = int(input("Enter Coding Score: "))
    comm_score = int(input("Enter Communication Score: "))

    student = {"math": math_score, "coding": coding_score, "communication": comm_score}
    career = model.recommend(student)

    print("Recommended Career Path:", career)

# -----------------------------
# GUI Execution
# -----------------------------
def run_gui():
    model = CareerModel()
    model.train(data)

    def recommend_career():
        try:
            math_score = int(entry_math.get())
            coding_score = int(entry_coding.get())
            comm_score = int(entry_comm.get())

            student = {"math": math_score, "coding": coding_score, "communication": comm_score}
            career = model.recommend(student)

            messagebox.showinfo("Career Recommendation", f"Recommended Career Path: {career}")
        except ValueError:
            messagebox.showerror("Input Error", "Please enter valid integer scores!")

    root = tk.Tk()
    root.title("AI Based Career Recommendation System")
    root.geometry("400x300")

    tk.Label(root, text="Math Score:").pack(pady=5)
    entry_math = tk.Entry(root)
    entry_math.pack(pady=5)

    tk.Label(root, text="Coding Score:").pack(pady=5)
    entry_coding = tk.Entry(root)
    entry_coding.pack(pady=5)

    tk.Label(root, text="Communication Score:").pack(pady=5)
    entry_comm = tk.Entry(root)
    entry_comm.pack(pady=5)

    tk.Button(root, text="Recommend Career", command=recommend_career).pack(pady=20)

    root.mainloop()

# -----------------------------
# Main Entry Point
# -----------------------------
if __name__ == "__main__":
    print("Choose Mode:")
    print("1. Command Line")
    print("2. GUI Window")
    choice = input("Enter choice (1/2): ")

    if choice == "1":
        run_cli()
    else:
        run_gui()
```
